## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
print("Number of unique content pages:", df["content_id"].nunique())

declining_rate = (df["trend_direction"].str.lower() == "down").mean()
print(f"\nShare of pages currently declining: {declining_rate:.1%}")

stale_pages = (df["days_since_last_update"] >= 180).mean()
print(f"Share of pages stale (not updated in 180+ days): {stale_pages:.1%}")

stale_and_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
print(f"Pages that are BOTH stale AND still getting real traffic: {stale_and_visible}")
```

The starter dataset contains 30,000 rows, 44 columns, and 30,000 unique content pages — one row per page, no duplicates. Over half (54.2%) of all pages are currently trending down, showing decline is common here, not an edge case — a real, sizeable prioritization problem. Only 0.6% of pages are stale (180+ days since update), meaning most content is fairly fresh, so staleness alone will be a weak signal in this dataset. Of that small stale group, 17 pages are both stale and still getting real traffic (500+ impressions in 90 days) — a concrete example of the kind of overlooked opportunity this lane is meant to catch.
